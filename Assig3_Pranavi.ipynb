```python
penguins = pd.read_csv("penguins_size.csv")
penguins.head()
```

```python
penguins['species'].unique()
```

```python
penguins = penguins.dropna()
```

```python
for sp in penguins['species'].unique():
    print(f"\nStatistical details of species: {sp}\n")
    print(
        penguins[penguins['species'] == sp]
        .describe(percentiles=[0.25, 0.5, 0.75])
    )
```

```python
iris = pd.read_csv("Iris.csv")
iris.head()
```

```python
iris.columns = iris.columns.str.strip().str.lower().str.replace(' ', '_')
iris.columns
```

```python
iris = iris.rename(columns={'class': 'species'})
```

```python
import pandas as pd
pd.__version__
```

```python
iris.columns
```

```python
iris_stats = iris.groupby('species').agg(
    mean_sepal_length=('sepallengthcm', 'mean'),
    std_sepal_length=('sepallengthcm', 'std'),
    min_sepal_length=('sepallengthcm', 'min'),
    max_sepal_length=('sepallengthcm', 'max')
)

iris_stats
```

```python
iris_percentiles = iris.groupby('species').quantile([0.25, 0.5, 0.75])
iris_percentiles
```
