## Dataset Description

The Iris dataset is an open-source dataset obtained from Kaggle:
https://www.kaggle.com/datasets/uciml/iris

The dataset contains 150 records of iris flowers.
Each record has the following attributes:
- SepalLengthCm
- SepalWidthCm
- PetalLengthCm
- PetalWidthCm
- Species

The objective of the dataset is to classify iris flowers into three species:
Setosa, Versicolor, and Virginica.

## Variable Description

- Id: Integer, unique identifier
- SepalLengthCm: Numeric, length of sepal
- SepalWidthCm: Numeric, width of sepal
- PetalLengthCm: Numeric, length of petal
- PetalWidthCm: Numeric, width of petal
- Species: Categorical variable representing flower species
